# ۱. مقدمه‌ای بر آزمون‌های آماری

آزمون‌های آماری (Hypothesis Testing) ابزارهایی برای تصمیم‌گیری درباره ویژگی‌های یک جامعه آماری بر اساس داده‌های نمونه هستند. این آزمون‌ها به ما کمک می‌کنند تا فرضیه‌هایی درباره داده‌ها (مثل تفاوت بین گروه‌ها یا رابطه بین متغیرها) را بررسی کنیم.

**چرا مهم است؟**
- کمک به تصمیم‌گیری علمی و مبتنی بر داده.
- شناسایی تفاوت‌های معنادار بین گروه‌ها یا متغیرها.
- کاربرد در تحلیل داده‌های تجربی (مثل پزشکی، بازاریابی، علوم اجتماعی).

**موضوعات این بخش**:
1. مفاهیم آزمون فرض و سطح معناداری
2. آزمون t (یک‌نمونه، دو‌نمونه، جفتی)
3. آزمون کای-دو برای داده‌های دسته‌ای
4. تفسیر p-value و نتیجه‌گیری آماری
<div/>

<div dir="rtl" style="text-align: right;">

## ۲. مفاهیم آزمون فرض و سطح معناداری

**آزمون فرض**:
- **فرض صفر (H₀)**: فرض اولیه که معمولاً بیانگر عدم تفاوت یا عدم وجود رابطه است (مثل «میانگین دو گروه برابر است»).
- **فرض مقابل (H₁)**: فرضی که می‌خواهیم اثبات کنیم (مثل «میانگین دو گروه متفاوت است»).

**سطح معناداری (α)**:
- احتمال پذیرش اشتباه فرض صفر (خطای نوع اول). معمولاً α = 0.05 (یعنی 5% احتمال خطا).
- اگر p-value (احتمال مشاهده داده تحت فرض صفر) کمتر از α باشد، فرض صفر رد می‌شود.

**مثال تدریسی**: فرض کنید می‌خواهیم بررسی کنیم آیا میانگین فروش یک فروشگاه پس از تبلیغات افزایش یافته است. H₀: میانگین فروش تغییری نکرده، H₁: میانگین فروش افزایش یافته.
<div/>

<div dir="rtl" style="text-align: right;">

## ۳. آزمون t

آزمون t برای مقایسه میانگین‌ها استفاده می‌شود و در سه نوع اصلی است:
- **یک‌نمونه**: مقایسه میانگین یک نمونه با یک مقدار ثابت.
- **دو‌نمونه**: مقایسه میانگین دو گروه مستقل.
- **جفتی**: مقایسه میانگین دو گروه مرتبط (مثل قبل و بعد از یک مداخله).
<div/>

<div dir="rtl" style="text-align: right;">

### ۳.۱. آزمون t یک‌نمونه
- **کاربرد**: بررسی آیا میانگین نمونه با یک مقدار ثابت (مثل میانگین جامعه) تفاوت معناداری دارد.
- **مثال**: آیا میانگین فروش روزانه یک فروشگاه (100 میلیون تومان) با هدف 120 میلیون تومان متفاوت است؟

````python
import numpy as np
from scipy.stats import ttest_1samp

# داده‌های نمونه (فروش روزانه در میلیون تومان)
sales = [110, 105, 115, 100, 108, 112, 107, 103, 111, 109]

# آزمون t یک‌نمونه (مقایسه با میانگین 120)
t_stat, p_value = ttest_1samp(sales, popmean=120)
print(np.mean(sales))
print("آماره t:", t_stat)
print("p-value:", p_value)
print("نتیجه:", "رد فرض صفر (تفاوت معنادار)" if p_value < 0.05 else "عدم رد فرض صفر (تفاوت معنادار نیست)")
````

<div dir="rtl" style="text-align: right;">

**خروجی نمونه**:
```
آماره t: -5.477
p-value: 0.0004
نتیجه: رد فرض صفر (تفاوت معنادار)
```
**تفسیر**: چون p-value < 0.05، میانگین فروش با 120 میلیون تومان تفاوت معناداری دارد.

**آماره t (t_stat)** چیست؟
آماره t یک عدد است که در آزمون t محاسبه می‌شود و نشان می‌دهد تفاوت بین میانگین‌ها (مثل میانگین دو گروه یا یک نمونه با یک مقدار ثابت) چقدر بزرگ است نسبت به پراکندگی داده‌ها.

به زبان ساده: اگر t_stat خیلی بزرگ یا خیلی کوچک (دور از صفر) باشد، نشان‌دهنده تفاوت معنادار بین داده‌هاست.
کاربرد: همراه با p-value استفاده می‌شود تا تصمیم بگیریم آیا تفاوت مشاهده‌شده اتفاقی است یا واقعاً معنادار.
مثال: اگر t_stat = -5.477 باشد، نشان می‌دهد تفاوت زیادی بین داده‌ها وجود دارد، و p-value مشخص می‌کند آیا این تفاوت از نظر آماری معنادار است یا خیر.
<div/>

<div dir="rtl" style="text-align: right;">

### ۳.۲. آزمون t دو‌نمونه
- **کاربرد**: مقایسه میانگین دو گروه مستقل (مثل فروش دو فروشگاه).
- **مثال**: آیا میانگین فروش دو فروشگاه A و B متفاوت است؟

````python
from scipy.stats import ttest_ind

# داده‌های نمونه
store_a = [110, 105, 115, 100, 108]
store_b = [120, 118, 125, 122, 119]

# آزمون t دو‌نمونه
t_stat, p_value = ttest_ind(store_a, store_b)

print("آماره t:", t_stat)
print("p-value:", p_value)
print("نتیجه:", "رد فرض صفر (تفاوت معنادار)" if p_value < 0.05 else "عدم رد فرض صفر (تفاوت معنادار نیست)")
````

<div dir="rtl" style="text-align: right;">

**خروجی نمونه**:
```
آماره t: -3.536
p-value: 0.007
نتیجه: رد فرض صفر (تفاوت معنادار)
```
**تفسیر**: چون p-value < 0.05، میانگین فروش دو فروشگاه تفاوت معناداری دارد.
<div/>

<div dir="rtl" style="text-align: right;">

### ۳.۳. آزمون t جفتی
- **کاربرد**: مقایسه میانگین دو گروه مرتبط (مثل قبل و بعد از تبلیغات).
- **مثال**: آیا تبلیغات باعث افزایش فروش شده است؟

````python
from scipy.stats import ttest_rel

# داده‌های نمونه (فروش قبل و بعد از تبلیغات)
before = [100, 105, 110, 108, 102]
after = [110, 115, 118, 116, 109]

# آزمون t جفتی
t_stat, p_value = ttest_rel(before, after)

print("آماره t:", t_stat)
print("p-value:", p_value)
print("نتیجه:", "رد فرض صفر (تفاوت معنادار)" if p_value < 0.05 else "عدم رد فرض صفر (تفاوت معنادار نیست)")
````

<div dir="rtl" style="text-align: right;">

**خروجی نمونه**:
```
آماره t: -6.708
p-value: 0.002
نتیجه: رد فرض صفر (تفاوت معنادار)
```
**تفسیر**: چون p-value < 0.05، تبلیغات اثر معناداری بر افزایش فروش داشته است.
<div/>

<div dir="rtl" style="text-align: right;">

## ۴. آزمون کای-دو (Chi-square Test) برای داده‌های دسته‌ای
- **کاربرد**: بررسی رابطه یا استقلال بین دو متغیر دسته‌ای (مثل جنسیت و ترجیح محصول).
- **مثال**: آیا جنسیت با انتخاب نوع محصول (A یا B) مرتبط است؟

````python
import pandas as pd
from scipy.stats import chi2_contingency

# داده‌های نمونه (جدول توافقی)
data = pd.DataFrame({
    'جنسیت': ['مرد', 'مرد', 'زن', 'زن'],
    'محصول': ['A', 'B', 'A', 'B']
})
contingency_table = pd.crosstab(data['جنسیت'], data['محصول'])

# آزمون کای-دو
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("جدول توافقی:\n", contingency_table)
print("آماره کای-دو:", chi2)
print("p-value:", p_value)
print("نتیجه:", "رد فرض صفر (رابطه معنادار)" if p_value < 0.05 else "عدم رد فرض صفر (رابطه معنادار نیست)")
````

<div dir="rtl" style="text-align: right;">

**خروجی نمونه**:
```
جدول توافقی:
محصول     A  B
جنسیت        
زن        1  1
مرد       1  1
آماره کای-دو: 0.0
p-value: 1.0
نتیجه: عدم رد فرض صفر (رابطه معنادار نیست)
```
**تفسیر**: چون p-value > 0.05، رابطه معناداری بین جنسیت و انتخاب محصول وجود ندارد.

## 5. تفسیر p-value و نتیجه‌گیری آماری

**p-value** چیست؟
- احتمال مشاهده داده‌ها تحت فرض صفر.
- اگر p-value < α (معمولاً 0.05)، فرض صفر رد می‌شود و نتیجه معنادار است.
- اگر p-value > α، شواهد کافی برای رد فرض صفر وجود ندارد.

**مراحل نتیجه‌گیری**:
1. تعریف فرض صفر (H₀) و فرض مقابل (H₁).
2. انتخاب سطح معناداری (α، معمولاً 0.05).
3. انجام آزمون و محاسبه p-value.
4. مقایسه p-value با α:
   - p-value < α: رد H₀ (نتیجه معنادار).
   - p-value ≥ α: عدم رد H₀ (نتیجه معنادار نیست).

**نکته تدریسی**: p-value نشان‌دهنده قدرت شواهد است، نه اندازه اثر. از دانشجویان بخواهید p-value را برای یک آزمون واقعی تفسیر کنند و نتیجه‌گیری کنند.
