## 1. Functions

A function is a block of code that runs when called. You can pass data (parameters) into a function, and it can return data as a result. Functions are defined using the `def` keyword.

### Key Concepts
- **Creating a Function**: Use `def` to define a function.
- **Arguments**: Pass information into functions.
- **Arbitrary Arguments (*args)**: Handle unknown numbers of arguments.
- **Keyword Arguments (kwargs)**: Pass arguments with key-value pairs.
- **Default Parameters**: Set default values for parameters.
- **Return Values**: Use `return` to output a value.
- **Recursion**: A function can call itself.

### Examples

```python
# Example 1: Basic Function
def greet():
    print("Hello from a function!")

greet()
```

```python
# Example 2: Function with Arguments
def print_name(fname):
    print(fname + " Smith")

print_name("Alice")
print_name("Bob")
```

```python
# Example 4: Keyword Arguments
def describe_person(name, age):
    print(f"{name} is {age} years old")

describe_person(age=25, name="Alice")
```

```python
# Example 6: Return Values
def multiply(x):
    return 5 * x

print(multiply(3))
print(multiply(10))
```

## 2. Classes

Python is an object-oriented programming language. A **class** is a blueprint for creating objects, which have properties and methods. Classes support **inheritance**, allowing one class to inherit attributes and methods from another.

### Key Concepts
- **Creating a Class**: Use the `class` keyword.
- **__init__()**: Constructor method to initialize object properties.
- **Inheritance**: A child class can inherit from a parent class.
- **super()**: Used to call the parent class’s methods.
- **Adding Properties and Methods**: Extend classes with new attributes and behaviors.

### Examples

```python
# Example 1: Creating a Class
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname
    
    def printname(self):
        print(self.firstname, self.lastname)

# Create an object
person = Person("John", "Doe")
person.printname()
```

```python
# Example 2: Inheritance
class Student(Person):
    pass  # Inherits all properties and methods from Person

student = Student("Mike", "Olsen")
student.printname()
```

```python
# Example 3: Child Class with __init__ and super()
class Student(Person):
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)
        self.graduationyear = year
    
    def welcome(self):
        print(f"Welcome {self.firstname} {self.lastname} to the class of {self.graduationyear}")

student = Student("Emma", "Watson", 2023)
student.welcome()
```

### Exercise: Classes
1. Create a class `Car` with properties `brand` and `model`, and a method to print the car details.
2. Create a child class `ElectricCar` that inherits from `Car` and adds a `battery_size` property and a method to display it.

Try solving these in the cell below:

```python
# Exercise Solutions
# 1. Car Class
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    def car_details(self):
        print(f"Car: {self.brand} {self.model}")

car = Car("Toyota", "Camry")
car.car_details()

# 2. ElectricCar Class
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size
    
    def battery_info(self):
        print(f"Battery Size: {self.battery_size} kWh")

ev = ElectricCar("Tesla", "Model 3", 75)
ev.car_details()
ev.battery_info()
```

## 3. Try Except

Exception handling in Python is done using the `try` and `except` blocks to handle errors gracefully. This prevents your program from crashing when an error occurs.

### Key Concepts
- **try**: Code that might cause an error is placed in the `try` block.
- **except**: Code to handle the error is placed in the `except` block.
- **else**: Executes if no exception occurs.
- **finally**: Executes regardless of whether an exception occurs.
- **Raising Exceptions**: Use `raise` to trigger exceptions manually.

### Examples

```python
# Example 1: Basic Try Except
try:
    print(x)
except:
    print("An error occurred: Variable x is not defined")
```

```python
# Example 2: Specific Exception
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed")
```

```python
# Example 4: Using else
try:
    x = 5
    print(x)
except:
    print("An error occurred")
else:
    print("No errors occurred")
```

```python
# Example 5: Using finally
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Error: Division by zero")
finally:
    print("This will always execute")
```

```python
# Example 6: Raising an Exception
x = -1
if x < 0:
    raise Exception("Error: Number cannot be negative")
```

### Exercise: Try Except
1. Write a function that attempts to divide two numbers and handles division by zero.
2. Create a function that converts a string to an integer and handles invalid input using try-except.

Try solving these in the cell below:

```python
# Exercise Solutions
# 1. Function to divide numbers with error handling
def divide_numbers(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")

divide_numbers(10, 2)
divide_numbers(10, 0)

# 2. Function to convert string to integer
def string_to_int(s):
    try:
        number = int(s)
        print(f"Converted number: {number}")
    except ValueError:
        print("Error: Invalid input, please provide a valid number")

string_to_int("123")
string_to_int("abc")
```

## 4. File Handling

File handling in Python allows you to create, read, write, and delete files. The key function is `open()`, which takes a filename and mode as parameters.

### Key Concepts
- **Modes**: 
  - `"r"`: Read (default)
  - `"a"`: Append
  - `"w"`: Write (overwrites)
  - `"x"`: Create (errors if file exists)
- **with Statement**: Automatically closes files.
- **Reading**: Use `read()`, `readline()`, or loop through lines.
- **Writing**: Use `write()` to add or overwrite content.

### Examples

**Note**: File handling examples may not work directly in some Jupyter environments due to file system restrictions. You can test these in a local Python environment.

```python
# Example 1: Reading a File
# Assume a file 'demofile.txt' exists with content: "Hello! Welcome to demofile.txt"
try:
    with open("demofile.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found. Please create 'demofile.txt' to test this example.")
```

```python
# Example 2: Writing to a File
try:
    with open("output.txt", "w") as f:
        f.write("This is a new file created by Python!")

    # Read the file to verify
    with open("output.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("Error: Could not write or read the file.")
```

```python
# Example 3: Appending to a File
try:
    with open("output.txt", "a") as f:
        f.write("\nAdding more content!")

    # Read the file to verify
    with open("output.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("Error: Could not append or read the file.")
```

### Exercise: File Handling
1. Write a function that creates a file named `log.txt` and writes a message to it, handling any file errors.
2. Write a function that reads `log.txt` and prints only the first line, handling file errors.

Try solving these in the cell below:

```python
# Exercise Solutions
# 1. Function to write to log.txt
def write_log(message):
    try:
        with open("log.txt", "w") as f:
            f.write(message)
        print("Successfully wrote to log.txt")
    except IOError:
        print("Error: Could not write to log.txt")

write_log("This is a log message.")

# 2. Function to read first line of log.txt
def read_first_line():
    try:
        with open("log.txt", "r") as f:
            print(f.readline().strip())
    except FileNotFoundError:
        print("File 'log.txt' not found.")
    except IOError:
        print("Error: Could not read from log.txt")

read_first_line()
```
