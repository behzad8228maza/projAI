## ۱. مقدمه‌ای بر مصورسازی داده‌ها

مصورسازی داده‌ها (Data Visualization) فرآیند نمایش داده‌ها به‌صورت بصری است تا الگوها، روندها، و روابط پنهان در داده‌ها به‌راحتی قابل درک شوند. مصورسازی به تحلیلگران کمک می‌کند تا:
- داده‌ها را خلاصه کنند.
- الگوها و روندها را شناسایی کنند.
- نتایج را به مخاطبان غیرفنی منتقل کنند.

**اهمیت**:
- انسان‌ها اطلاعات بصری را سریع‌تر درک می‌کنند.
- کمک به شناسایی داده‌های پرت، توزیع‌ها، و روابط بین متغیرها.
- ابزارهای اصلی: **Matplotlib** (برای نمودارهای پایه) و **Seaborn** (برای نمودارهای زیباتر و پیچیده‌تر).

**مثال تدریسی**: مقایسه فروش ماهانه یک شرکت با نمودار میله‌ای یا بررسی توزیع سنی مشتریان با هیستوگرام.

## ۲. نمودارهای پایه با Matplotlib

**Matplotlib** یک کتابخانه قدرتمند پایتون برای مصورسازی داده‌هاست. در این بخش، نمودارهای خطی، میله‌ای، 
دایره‌ای، و هیستوگرام را بررسی می‌کنیم.

<div/>

<div dir="rtl" style="text-align: right;">

### ۲.۱. نمودار خطی (Line Plot)
- **کاربرد**: نمایش روندهای پیوسته (مثل تغییرات دما در طول زمان).
- **ویژگی‌ها**: برای داده‌های پیوسته یا زمانی مناسب است.

<div/>

```python
import matplotlib.pyplot as plt
import numpy as np

# داده‌های نمونه
months = np.array(['فروردین', 'اردیبهشت', 'خرداد', 'تیر', 'مرداد', 'شهریور'])
sales = np.array([100, 120, 130, 150, 140, 160])

plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker='o', color='blue', linestyle='--')
plt.title('فروش ماهانه شرکت')
plt.xlabel('ماه')
plt.ylabel('فروش (میلیون تومان)')
plt.grid(True)
plt.show()
```

<div dir="rtl" style="text-align: right;">

**خروجی**: نمودار خطی که روند افزایشی فروش را نشان می‌دهد.
<div/>

<div dir="rtl" style="text-align: right;">

### ۲.۲. نمودار میله‌ای (Bar Plot)
- **کاربرد**: مقایسه مقادیر گسسته (مثل فروش محصولات مختلف).
- **ویژگی‌ها**: برای داده‌های دسته‌ای مناسب است.

<div/>

```python
import matplotlib.pyplot as plt

# داده‌های نمونه
products = ['محصول A', 'محصول B', 'محصول C']
sales = [200, 150, 300]

plt.figure(figsize=(7, 5))
plt.bar(products, sales, color='green')
plt.title('مقایسه فروش محصولات')
plt.xlabel('محصول')
plt.ylabel('فروش (میلیون تومان)')
plt.show()
```

<div dir="rtl" style="text-align: right;">

**خروجی**: نمودار میله‌ای که محصول C بیشترین فروش را دارد.
<div/>

<div dir="rtl" style="text-align: right;">

### ۲.۳. نمودار دایره‌ای (Pie Chart)
- **کاربرد**: نمایش نسبت یا درصد دسته‌ها (مثل سهم بازار).
- **ویژگی‌ها**: برای داده‌های درصدی با تعداد دسته‌های کم مناسب است.
<div/>

```python
import matplotlib.pyplot as plt

# داده‌های نمونه
labels = ['محصول A', 'محصول B', 'محصول C']
sizes = [35, 25, 40]
colors = ['#ff9999', '#66b3ff', '#99ff99']

plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('سهم بازار محصولات')
plt.axis('equal')
plt.show()
```

<div dir="rtl" style="text-align: right;">

**خروجی**: نمودار دایره‌ای که نشان می‌دهد محصول C با ۴۰٪ بیشترین سهم را دارد.
<div/>

<div dir="rtl" style="text-align: right;">

### ۲.۴. هیستوگرام (Histogram)
- **کاربرد**: نمایش توزیع داده‌ها (مثل توزیع سنی مشتریان).
- **ویژگی‌ها**: برای داده‌های پیوسته و بررسی شکل توزیع مناسب است.
<div/>

```python
import matplotlib.pyplot as plt
import numpy as np

# داده‌های نمونه
ages = np.random.normal(30, 10, 1000)  # توزیع سنی فرضی

plt.figure(figsize=(8, 5))
plt.hist(ages, bins=20, color='purple', edgecolor='black')
plt.title('توزیع سنی مشتریان')
plt.xlabel('سن')
plt.ylabel('تعداد')
plt.grid(True, alpha=0.3)
plt.show()
```

<div dir="rtl" style="text-align: right;">

**خروجی**: هیستوگرام با توزیع نرمال حول سن ۳۰ سال.

30 → میانگین (mean) داده‌ها

10 → انحراف معیار (standard deviation)، یعنی داده‌ها معمولاً ±۱۰ واحد اطراف میانگین پراکنده‌اند

1000 → تعداد داده‌ها که تولید می‌کنیم

**نکته تدریسی**: از دانشجویان بخواهید داده‌های خود (مثل نمرات امتحان) را با این نمودارها نمایش دهند.

## ۳. نمودارهای پیشرفته با Seaborn

**Seaborn** یک کتابخانه مبتنی بر Matplotlib است که نمودارهای زیباتر و پیچیده‌تر را با کد کمتر تولید می‌کند.
<div/>

<div dir="rtl" style="text-align: right;">

### ۳.۱. نمودار جعبه‌ای (Boxplot)
- **کاربرد**: نمایش توزیع داده‌ها، میانه، چارک‌ها، و داده‌های پرت.
- **ویژگی‌ها**: برای مقایسه توزیع بین چند گروه مناسب است.
<div/>

```python
import seaborn as sns
import pandas as pd
import numpy as np

# داده‌های نمونه
data = pd.DataFrame({
    'گروه': ['A']*100 + ['B']*100 + ['C']*100,
    'مقدار': np.concatenate([
        np.random.normal(50, 10, 100),  # گروه A    mean=50, std=10
        np.random.normal(60, 15, 100),  # گروه B
        np.random.normal(55, 12, 100)   # گروه C
    ])
})

plt.figure(figsize=(8, 5))
sns.boxplot(x='گروه', y='مقدار', data=data, palette='Set2')
plt.title('مقایسه توزیع مقادیر بین گروه‌ها')
plt.show()
data
```

<div dir="rtl" style="text-align: right;">

**خروجی**: نمودار جعبه‌ای که میانه، چارک‌ها، و پرت‌های هر گروه را نشان می‌دهد.
<div/>

<div dir="rtl" style="text-align: right;">

### ۳.۲. نمودار پراکندگی (Scatter Plot)
- **کاربرد**: نمایش رابطه بین دو متغیر (مثل قد و وزن).
- **ویژگی‌ها**: برای بررسی همبستگی یا الگوهای خوشه‌ای مناسب است.
<div/>

```python
import seaborn as sns
import pandas as pd
import numpy as np

# داده‌های نمونه
np.random.seed(42)
data = pd.DataFrame({
    'قد': np.random.normal(170, 10, 100),
    'وزن': np.random.normal(70, 15, 100)
})

plt.figure(figsize=(8, 5))
sns.scatterplot(x='قد', y='وزن', data=data, color='blue')
plt.title('رابطه قد و وزن')
plt.show()
```

<div dir="rtl" style="text-align: right;">

**خروجی**: نمودار پراکندگی که رابطه احتمالی بین قد و وزن را نشان می‌دهد.
<div/>

<div dir="rtl" style="text-align: right;">

## ۴. تحلیل دیداری داده‌ها و کشف الگوها

تحلیل دیداری داده‌ها فرآیند شناسایی الگوها، روندها، و ناهنجاری‌ها از طریق نمودارهاست. مراحل اصلی:
1. **تعیین هدف**: چه چیزی می‌خواهید کشف کنید؟ (مثل روند فروش یا شناسایی پرت‌ها)
2. **انتخاب نمودار مناسب**:
   - روندها: نمودار خطی.
   - مقایسه: نمودار میله‌ای یا جعبه‌ای.
   - توزیع: هیستوگرام یا جعبه‌ای.
3. **تحلیل الگوها**:
   - **پراکندگی**: شناسایی خوشه‌ها یا پرت‌ها (مثل مشتریان غیرعادی).
   - **روندها**: افزایش یا کاهش در طول زمان.
   - **همبستگی**: روابط قوی یا ضعیف بین متغیرها.
4. **نتیجه‌گیری**: تبدیل مشاهدات بصری به تصمیمات (مثل بهبود استراتژی فروش).

**مثال عملی**:
فرض کنید داده‌های فروش یک فروشگاه دارید:
- **هیستوگرام**: توزیع مبالغ خرید مشتریان (آیا اکثر خریدها کوچک هستند؟).
- **نمودار خطی**: روند فروش در ۱۲ ماه (آیا فصلی است؟).
- **پراکندگی**: رابطه بین تبلیغات و فروش (آیا تبلیغات اثر دارد؟).
<div/>

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# داده‌های نمونه
np.random.seed(42)
data = pd.DataFrame({
    'ماه': np.repeat(['فروردین', 'اردیبهشت', 'خرداد', 'تیر'], 25),
    'فروش': np.random.normal(150, 20, 100),
    'تبلیغات': np.random.normal(50, 10, 100)
})

# نمودارهای مختلف
plt.figure(figsize=(15, 10))

# هیستوگرام
plt.subplot(2, 2, 1)
sns.histplot(data['فروش'], bins=20, kde=True, color='purple')
plt.title('توزیع فروش')

# نمودار خطی
plt.subplot(2, 2, 2)
sns.lineplot(x='ماه', y='فروش', data=data, marker='o')
plt.title('روند فروش ماهانه')

# پراکندگی
plt.subplot(2, 2, 3)
sns.scatterplot(x='تبلیغات', y='فروش', data=data, color='blue')
plt.title('رابطه تبلیغات و فروش')
```


<div dir="rtl" style="text-align: right;">
**خروجی**: سه نمودار که توزیع، روند و رابطه را نشان می‌دهند.

**نکات کشف الگوها**:
- **هیستوگرام**: بررسی چولگی یا پرت‌ها.
- **پراکندگی**: شناسایی خوشه‌ها یا روابط خطی/غیرخطی.
